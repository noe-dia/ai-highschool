### ✅❌ Les variables booléennes et les opérateurs logiques 

Les variables booléennes ne peuvent prendre que deux valeurs, soit `True` soit `False`. 

Les variables booléennes font intervenir des opérateurs dits logiques: `and`, `or` et `not`

Opérateur `&` ou `and`
| `and` | True | False |
|:-----|----:|:-----:|
| True | True | False |
| False | False| False |

Opérateur `|` ou `or`
| `or` | True | False |
|:-----|----:|:-----:|
| True | True | True |
| False | True| False |

L'opérateur `not` renvoie l'opposé du booléen qu'on lui donne en entrée. Donc, si j'écris,
```python
x = not True
```
x contient le booléen `False`

Ces variables nous sera notamment utile quand on écrira du code qui sera dépendant de condition. 

````python
print(True and False)
print(True & False) # une autre notation possible.
print(True or False)
print(True | False) # une autre notation possible. 
print(not True)
````
