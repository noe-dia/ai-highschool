##### Qu'est-ce qu'un objet ?
Python est un langage de programmation qui se base sur les objets. Le langage tout entier est construit autour de cette notion et nous avons pu rencontrer des objets depuis le début du cours. Les différents types de variables sont des objets par exemple auxquelles sont associés des attributs et des méthodes spécifiques.

La programmation orientée objet consiste donc à repenser la manière dont on construit son code et à créer nos propres objets. 

Notons que les attributs et méthodes de chaque objet peuvent être identifiés grâce à la fonction `dir`. 

```python
dir("3")
```

##### Définir ses propres classes

Comme il est intéressant pour nous de définir nos propres fonctions, il sera aussi intéressant de définir nos propres classes. 

Prenons comme exemple une classe `Cat` qui contient les informations relatives à un animal de compagnie, un chat par exemple. On peut créer une classe chat comme suit, 

```python
class Cat:
    ...
```

Maintenant, on va vouloir spécifier certaines caractéristiques du chat. Lorsque la classe sera appelée la première fois et sera initialisée, on veut accéder aux caractéristiques assez rapidement. 

On peut créer pour ça une méthode `__init__`.

```python
class Cat:
    def __init__(self, mass, name, color):
        self.mass = mass
        self.name = name 
        self.color = color
```

On note l'utilisation de `self` ici qui réfère à la classe. `__init__` permet ici de créer des attributs auxquels on va avoir accès en tout temps une fois la classe définie. Par exemple, 

```python
mass = 10
name = "Oscar"
color = "black"
cat = Cat(mass = mass, name = name, color = color)

print("Masse", cat.mass)
print("Name", cat.name)
print("Color", cat.color)
```

On va pouvoir également ajouter des _méthodes_ à notre classe, c'est-à-dire des fonctions auxquelles on a accès en tout temps avec la classe.

```python
class Cat:
    def __init__(self, mass, name, color):
        self.mass = mass
        self.name = name 
        self.color = color

    def hello(self):
        print("Hello, I am", self.name, "the cat !")

cat = Cat(mass = mass, name = name, color = color)
cat.hello()
```

On peut ajouter d'autres méthodes à volonté et les appeler 

```python
class Cat:
    def __init__(self, mass, name, color):
        self.mass = mass
        self.name = name 
        self.color = color

    def hello(self):
        print("Hello, I am", self.name, "the cat !")

    def pet(self):
        print("*Ronronne*")
cat = Cat(mass = mass, name = name, color = color)
cat.hello()
cat.pet()
```

Une autre chose qui va être intéressante est l'inhéritance en programmation. On peut créer une classe générale `Pet` qui va regrouper des méthodes utiles pour des sous-classes comme `Cat` ou `Dog` par exemple

```python
class Pet:
    def __init__(self, mass, name, color):
        self.mass = mass
        self.name = name
        self.color = color
        self.type = None

    def hello(self):
        print("Hello, I am", self.name, "the", self.type, "!") 

    def caracteristic(self):
        print("I weigh", self.mass, "kg and I am", self.color)
```

```python
class Cat(Pet):
    def __init__(self, mass, name, color):
        super().__init__(mass, name, color)
        self.type = "cat"

class Dog(Pet):
    def __init__(self, mass, name, color):
        super().__init__(mass, name, color)
        self.type = "dog"

cat = Cat(mass = 10, name = "Garfield", color = "red")
dog = Dog(mass = 10, name = "Carlos", color = "grey")
cat.caracteristic()
cat.hello()
```

Une autre méthode pratique est la méthode `__call__` qui va nous permettre d'utiliser une classe comme une fonction et de l'appeler. On peut par exemple changer notre classe `Pet` ici. 

```python
class Pet:
    def __init__(self, mass, name, color):
        self.mass = mass
        self.name = name
        self.color = color
        self.type = None

    def hello(self):
        print("Hello, I am", self.name, "the", self.type, "!") 

    def caracteristic(self):
        print("I weigh", self.mass, "kg and I am", self.color)

    def __call__(self, pet):
        print(self.name, "the", self.type, "ate", pet.name, "the", pet.type, "!!!")

class Cat(Pet):
    def __init__(self, mass, name, color):
        super().__init__(mass, name, color)
        self.type = "cat"

class Dog(Pet):
    def __init__(self, mass, name, color):
        super().__init__(mass, name, color)
        self.type = "dog"
```

```python
cat = Cat(mass = 10, name = "Garfield", color = "red")
dog = Dog(mass = 10, name = "Carlos", color = "grey")

dog(cat)
```
